```python
data <- c(30, 40, 50, 60, 80, 90, 100)
print("Dataset")
print(data)
mean_value <- mean(data)
print(paste("Mean =", mean_value))
median_value <- median(data)
print(paste("Median =", median_value))
mode_function <- function(x) {
  uniq_values <- unique(x)
  uniq_values[which.max(tabulate(match(x, uniq_values)))]}
mode_value <- mode_function(data)
print(paste("Mode =", mode_value))
max_value <- max(data)
print(paste("Maximum =", max_value))
min_value <- min(data)
print(paste("Minimum =", min_value))
variance_value <- var(data)
print(paste("Variance =", variance_value))
sd_value <- sd(data)
print(paste("Standard Deviation =", sd_value))
summary(data)
barplot(data)
hist(data)
pie(data)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv("/content/gym_members_exercise_tracking_synthetic_data.csv")
ages = data["Age"]
workout_hours = data["Session_Duration (hours)"]
calories_burned = data["Calories_Burned"]
mem_names = []
for i in range(len(data)):
    mem_names.append("Member_" + str(i + 1))
print("FITNESS MEMBERS DATA\n")
for i in range(len(mem_names)):
    print(mem_names[i],
          "| Age:", ages[i],
          "| Workout Hours:", workout_hours[i],
          "| Calories Burned:", calories_burned[i])
avg_workout = workout_hours.mean()
print("\nAverage Workout Hours:", avg_workout)
max_calories = calories_burned.max()
print("Maximum Calories Burned:", max_calories)
min_calories = calories_burned.min()
print("Minimum Calories Burned:", min_calories)
median_calories = calories_burned.median()
print("Median Calories Burned:", median_calories)
std_dev = calories_burned.std()
print("Standard Deviation:", std_dev)
active_index = calories_burned.idxmax()
print("\nMOST ACTIVE PERSON")
print("Name:", mem_names[active_index])
print("Calories Burned:", calories_burned[active_index])
print("Workout Hours:", workout_hours[active_index])
lazy_index = calories_burned.idxmin()
print("\nLAZY PERSON")
print("Name:", mem_names[lazy_index])
print("Calories Burned:", calories_burned[lazy_index])
print("Workout Hours:", workout_hours[lazy_index])
plt.figure(figsize=(12, 5))
plt.bar(mem_names[:20], calories_burned[:20])
plt.xlabel("Member Names")
plt.ylabel("Calories Burned")
plt.title("Fitness Analysis - Bar Graph")
plt.xticks(rotation=90)
plt.show()
plt.figure(figsize=(8, 8))
plt.pie(calories_burned[:10],
        labels=mem_names[:10],
        autopct='%1.1f%%')
plt.title("Calories Burned Distribution")
plt.show()
plt.figure(figsize=(10, 5))
plt.scatter(workout_hours, calories_burned)
plt.xlabel("Workout Hours")
plt.ylabel("Calories Burned")
plt.title("Workout Hours vs Calories Burned")
plt.show()
```
